```python
import numpy as np
import matplotlib.pyplot as plt

#Task 1 : Define Activation Functions

#Sigmoid Function
def sigmoid(x):
  return 1/(1+ np.wxp(-x))

#ReLU Function
def relu(x):
  return np.maximum(0,x)

#Tanh Function
def tanh(x):
  return np.tanh(x)
```

```python
# Task 2: Generate Outputs

# Input values
x_values = np.array([-5,-3,-1,0,1,3,5])

# Apply activation Function
sigmoid_outputs = sigmoid(x_values)
relu_outputs = relu(x_values)
tanh_outputs = tanh(x_values)

# Display the outputs
print("Input Values:")
print(x_values)

print("\nSigmoid Outputs:")
print(sigmoid_outputs)

print("\nReLU Outputs:")
print(relu_outputs)

print("\nTanh Outputs:")
print(tanh_outputs)
```

```python
#Task 3 : Visualize and Compare
#Generate continuous input values for plotting

x = np.linspace(-5 , 5 , 100)

#Calculate outputs
y_sigmoid = sigmoid(x)
y_relu = relu(x)
y_tanh = tanh(x)

#Plot the outputs

plt.figure(figsize=(10,6))

plt.plot(x,y_sigmoid , label = 'Sigmoid')
plt.plot(x,y_relu , label = 'ReLU')
plt.plot(x,y_tanh , label = 'Tanh')

plt.xlabel('Input Values')
plt.ylabel('Output Values')
plt.title('Comparison of Activation Functions')

plt.grid(True)
plt.legend()
plt.show()
```
